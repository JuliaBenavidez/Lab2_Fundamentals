```python
# EXERCISE 1: SENSOR MONITORING

ENGINEER = "Benavidez"
STUDENT_ID = "TUPM-26-1835"
SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")

# Generated sensor data

sensor_data = []

for i in range(SEED_NUM + 3):
    reading = (ord(ENGINEER[i % len(ENGINEER)]) + SEED_NUM + i) % 101
    sensor_data.append(reading)

print("Generated Sensor Data:")
print(sensor_data)

# Validate sensor data

valid_readings = []
invalid_readings = []

for reading in sensor_data:
    try:
        value = int(reading)

        if 0 <= value <= 100:
            valid_readings.append(value)
        else:
            invalid_readings.append(value)

    except ValueError:
        invalid_readings.append(reading)

print("Valid Readings:", valid_readings)
print("Invalid Readings:", invalid_readings)

# Classify valid data

classification_results = []

for value in valid_readings:

    if value <= 30:
        classification = "LOW"

    elif value <= 70:
        classification = "NORMAL"

    else:
        classification = "HIGH"

    classification_results.append(
        f"{value} -> {classification}"
    )

print("Classification Results:")

for result in classification_results:
    print(result)

# Final assessment

print("=== SENSOR MONITORING ===")
print(f"Generated Sensor Data: {sensor_data}")
print(f"Valid Readings: {valid_readings}")
print(f"Invalid Readings: {invalid_readings}")
print(f"Classification Results: {classification_results}")

print("\nExecution Log:")
print(f"Total Readings: {len(sensor_data)}")
print(f"Valid: {len(valid_readings)}")
print(f"Invalid: {len(invalid_readings)}")
```

```python
# EXERCISE 2: TEXT SIGNAL ANALYSIS

ENGINEER = "Benavidez"
STUDENT_ID = "TUPM-26-1835"
SEED_NUM = int(STUDENT_ID[-1])

# Generate a unique signal
signal = ENGINEER + str(SEED_NUM)

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
print(f"Generated Signal: {signal}")

# CHARACTER-BY-CHARACTER ANALYSIS

letters = 0
numbers = 0
spaces = 0
special_characters = 0

print("=== CHARACTER ANALYSIS ===")

for character in signal:

    if character.isalpha():
        character_type = "LETTER"
        letters += 1

    elif character.isdigit():
        character_type = "NUMBER"
        numbers += 1

    elif character.isspace():
        character_type = "SPACE"
        spaces += 1

    else:
        character_type = "SPECIAL CHARACTER"
        special_characters += 1

    print(f"{character} -> {character_type}")

    # ASCII ANALYSIS

print("=== ASCII VALUES ===")

for character in signal:
    print(f"{character} -> ASCII {ord(character)}")

# FINAL DIAGNOSTIC REPORT

print("=== DIAGNOSTIC REPORT ===")

print(f"Generated Signal: {signal}")
print(f"Letters: {letters}")
print(f"Numbers: {numbers}")
print(f"Spaces: {spaces}")
print(f"Special Characters: {special_characters}")

print("\nExecution Log:")
print(f"Characters Processed: {len(signal)}")
print(f"Letters Detected: {letters}")
print(f"Numbers Detected: {numbers}")
print(f"Spaces Detected: {spaces}")
print(f"Special Characters Detected: {special_characters}")

# FINAL DIAGNOSTIC REPORT

# Determine signal classification
if letters > 0 and numbers > 0 and spaces == 0 and special_characters == 0:
    signal_classification = "ALPHANUMERIC"
elif letters > 0 and numbers == 0 and spaces == 0 and special_characters == 0:
    signal_classification = "LETTERS ONLY"
elif numbers > 0 and letters == 0 and spaces == 0 and special_characters == 0:
    signal_classification = "NUMBERS ONLY"
else:
    signal_classification = "MIXED"

print("=== DIAGNOSTIC REPORT ===")

print(f"Generated Signal: {signal}")
print(f"Processed Signal: {len(signal)} characters")
print(f"Character Analysis: {letters} letters, {numbers} numbers, {spaces} spaces, {special_characters} special characters")
print(f"Signal Classification: {signal_classification}")

print("\nExecution Log:")
print(f"Signal processed: {signal}")
print(f"Characters processed: {len(signal)}")
print(f"Letters detected: {letters}")
print(f"Numbers detected: {numbers}")
print(f"Spaces detected: {spaces}")
print(f"Special characters detected: {special_characters}")
print(f"Classification: {signal_classification}")
```

```python
# EXERCISE 3: PASSWORD AUTHENTICATION

ENGINEER = "Benavidez"
STUDENT_ID = "TUPM-26-1835"
SEED_NUM = int(STUDENT_ID[-1])

generated_password = str(SEED_NUM - 4 + 10) + ENGINEER

print(f"Generated Password: {generated_password}")

# ATTEMPT LIMIT

attempt_limit = (SEED_NUM % 3) + 3

print(f"Attempt Limit: {attempt_limit}")

# PASSWORD AUTHENTICATION

attempts_made = 0
access_granted = False
execution_log = []

while attempts_made < attempt_limit:

    entered_password = input("Enter Password: ")

    attempts_made += 1

    if entered_password == generated_password:

        access_granted = True

        execution_log.append(
            f"Attempt {attempts_made}: Access Granted"
        )

        print("Access Granted!")
        break

    else:

        execution_log.append(
            f"Attempt {attempts_made}: Access Denied"
        )

        print("Incorrect Password.")

if access_granted:
    final_state = "UNLOCKED"
else:
    final_state = "LOCKED"
    print("Maximum attempts reached.")

# AUTHENTICATION REPORT

print("=== AUTHENTICATION REPORT ===")

print(f"Generated Password: {generated_password}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Attempts Made: {attempts_made}")

if access_granted:
    print("Access Result: GRANTED")
else:
    print("Access Result: DENIED")

print(f"Final System State: {final_state}")

print("\nExecution Log:")

for log in execution_log:
    print(log)
```
